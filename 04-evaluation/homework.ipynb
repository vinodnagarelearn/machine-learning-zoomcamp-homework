```python
df = pd.read_csv('hw-week-4.csv')
df 
```

```python
df.isnull().sum()
```

```python
#replace the space by underscore in 
df.columns = df.columns.str.lower().str.replace(' ', '_')

#get names of categorical variables
categorical_columns = list(df.dtypes[df.dtypes == 'str'].index)

#in categorical variable replace space by underscore
for c in categorical_columns:
    df[c] = df[c].str.lower().str.replace(' ', '_')
categorical_columns
```

```python
#Fill zero in numerical variable for NAN
df.lead_score = pd.to_numeric(df.lead_score, errors='coerce')
df.lead_score = df.lead_score.fillna(0)

df.annual_income = pd.to_numeric(df.annual_income, errors='coerce')
df.annual_income = df.annual_income.fillna(0)

#Fill NA in numerical categorical for NAN
df[categorical_columns] = df[categorical_columns].fillna('NA')

df.isnull().sum()
```

```python
#Split the data into 3 parts with these exact calls:

df_full_train, df_test = train_test_split(df, test_size=0.2, random_state=1)
df_train, df_val = train_test_split(df_full_train, test_size=0.25, random_state=1 )

len(df), len(df_full_train), len(df_test), len(df_train), len(df_val)
```
